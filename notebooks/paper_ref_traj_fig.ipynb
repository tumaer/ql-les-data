```python
import h5py
import matplotlib.pyplot as plt
import numpy as np
```

```python
path = "../dataset_kolm/raw/2D_KOLM_4096_140kevery1/traj_15/u_512_04500_burnin.npy"
u = np.load(path) # shape (2, 512, 512)
speed = np.linalg.norm(u, axis=0)
fig, ax = plt.subplots(figsize=(5, 5), layout="constrained")
im = ax.imshow(speed, origin="lower", cmap="viridis")
ax.set_xticks([])
ax.set_yticks([])
fig.savefig('filtered_velocity.png', dpi=300)
```

```python
path = "../dataset_kolm/datasets/2D_KOLM_4096_140kevery1/test.h5"
data = h5py.File(path, 'r')
r0 = np.asarray(data["/00000/position"][0])
data.close()
fig, ax = plt.subplots(figsize=(5, 5), layout='constrained')
ax.scatter(r0[:, 0], r0[:, 1], s=40)
ax.set_xlim(0, 2*np.pi/2)
ax.set_ylim(0, 2*np.pi/2)
ax.set_aspect('equal')
ax.set_xticks([])
ax.set_yticks([])
fig.savefig('particles.png', dpi=600)
```
