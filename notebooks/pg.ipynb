```python
r = np.random.randn(1000000)
```

```python
# Explore how the turbulent Mach number from Tian et al. 2023 behaves
stds = np.array([0.05, 0.1, 0.2])
a = np.array([1 + i * r for i in stds])
```

```python
a.mean(axis=1), a.std(axis=1)
```

```python
a_rms = np.sqrt(np.mean(a**2, axis=1))
a_rms, (a_rms-1)/(stds**2)
```
